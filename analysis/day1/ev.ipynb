```python
import matplotlib.pyplot as plt
plt.figure()
df.boxplot(column="Energy Consumed (kWh)")
plt.show()
```

```python
plt.figure()
df.boxplot(column="Charging Cost (USD)", by="Vehicle Age (years)")
plt.xlabel("Vehicle Age(years)")
plt.ylabel("Charging Cost (USD)")
plt.title("Charging Cost by Vehicle Age")
plt.suptitle("")  # remove auto title
plt.show()
```
